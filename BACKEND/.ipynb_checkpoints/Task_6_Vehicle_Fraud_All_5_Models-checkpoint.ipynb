# Task 6 — Vehicle Insurance Fraud Model Comparison

This notebook uses the **same 23 user-input fields as the current FastAPI `main.py`**.

Models compared:
1. Logistic Regression
2. Decision Tree
3. Random Forest
4. AdaBoost
5. Gradient Boosting

The final model is selected automatically using **genuine test-set accuracy** and saved as `vehicle_fraud_final_model.pkl`.

> Accuracy is not artificially increased or forced to a target percentage.


```python
# CELL 1 — IMPORT LIBRARIES

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
```

```python
# CELL 2 — LOAD DATASET

CSV_PATH = Path("insurance_fraud_data.csv")

df = pd.read_csv(CSV_PATH)

print("=" * 70)
print("STEP 1 - ORIGINAL DATASET")
print("=" * 70)
print("Shape:", df.shape)
```

```python
# CELL 3 — CLEAN TARGET

TARGET = "fraud reported"

df[TARGET] = df[TARGET].replace({
    "N": 0,
    "Y": 1
})

df[TARGET] = pd.to_numeric(df[TARGET], errors="coerce")

df = df.dropna(subset=[TARGET]).copy()
df[TARGET] = df[TARGET].astype(int)

print("=" * 70)
print("STEP 3 - TARGET")
print("=" * 70)
print(df[TARGET].value_counts())
print()
print(df[TARGET].value_counts(normalize=True) * 100)
```

```python
# CELL 4 — EXACT 23 API INPUT COLUMNS
# These are the fields currently accepted by main.py.

INPUT_COLUMNS = [
    "age_of_driver",
    "safety_rating",
    "annual_income",
    "high_education",
    "address_change",
    "property_status",
    "claim_date",
    "claim_day_of_week",
    "accident_site",
    "past_num_of_claims",
    "witness_present",
    "liab_prct",
    "channel",
    "police_report",
    "age_of_vehicle",
    "vehicle_category",
    "vehicle_price",
    "total_claim",
    "injury_claim",
    "policy deductible",
    "annual premium",
    "days open",
    "form defects"
]

missing_columns = [
    col for col in INPUT_COLUMNS + [TARGET]
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required dataset columns: {missing_columns}"
    )

X = df[INPUT_COLUMNS].copy()
y = df[TARGET].copy()

print("=" * 70)
print("STEP 4 - EXACT API FEATURES")
print("=" * 70)
print("Feature count:", len(INPUT_COLUMNS))
print(INPUT_COLUMNS)
```

```python
# CELL 5 — FEATURE ENGINEERING

def create_features(data):
    data = data.copy()

    # Convert fields that may arrive as strings in the dataset/API
    numeric_columns = [
        "witness_present",
        "age_of_vehicle",
        "injury_claim"
    ]

    for column in numeric_columns:
        data[column] = pd.to_numeric(
            data[column],
            errors="coerce"
        )

    # Date conversion
    data["claim_date"] = pd.to_datetime(
        data["claim_date"],
        errors="coerce"
    )

    # Date-derived features
    data["claim_year"] = data["claim_date"].dt.year
    data["claim_month"] = data["claim_date"].dt.month
    data["claim_day"] = data["claim_date"].dt.day
    data["claim_day_number"] = data["claim_date"].dt.dayofweek
    data["claim_week"] = (
        data["claim_date"]
        .dt.isocalendar()
        .week
        .astype(float)
    )

    # Ratio features
    data["claim_vehicle_ratio"] = (
        data["total_claim"] /
        (data["vehicle_price"] + 1)
    )

    data["claim_premium_ratio"] = (
        data["total_claim"] /
        (data["annual premium"] + 1)
    )

    data["injury_claim_ratio"] = (
        data["injury_claim"] /
        (data["total_claim"] + 1)
    )

    # Original date is no longer required
    data = data.drop(
        columns=["claim_date"]
    )

    return data


X = create_features(X)

print("=" * 70)
print("STEP 5 - FEATURE ENGINEERING")
print("=" * 70)
print("Feature shape:", X.shape)
print("Features:")
print(list(X.columns))
```

```python
# CELL 6 — IDENTIFY NUMERICAL AND CATEGORICAL FEATURES

numerical_features = X.select_dtypes(
    include=["int64", "float64", "Int64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("=" * 70)
print("STEP 6 - FEATURE TYPES")
print("=" * 70)

print("Numerical features:")
print(numerical_features)

print()
print("Categorical features:")
print(categorical_features)
```

```python
# CELL 7 — TRAIN / TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("=" * 70)
print("STEP 7 - TRAIN TEST SPLIT")
print("=" * 70)
print("Training:", len(X_train))
print("Testing :", len(X_test))
```

```python
# CELL 8 — PREPROCESSING

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
# CELL 9 — DEFINE ALL 5 MODELS

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features="sqrt",
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        min_samples_leaf=5,
        random_state=42
    )
}
```

```python
# CELL 10 — TRAIN ALL 5 MODELS

results = []
trained_pipelines = {}

for model_name, model_instance in models.items():

    print(f"Training: {model_name}")

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model_instance)
    ])

    pipeline.fit(
        X_train,
        y_train
    )

    predictions = pipeline.predict(X_test)

    probabilities = pipeline.predict_proba(
        X_test
    )[:, 1]

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        probabilities
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })

    trained_pipelines[model_name] = pipeline
```

```python
# CELL 11 — MODEL COMPARISON

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("=" * 70)
print("STEP 11 - MODEL COMPARISON")
print("=" * 70)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy": "{:.4f}".format,
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "F1 Score": "{:.4f}".format,
            "ROC-AUC": "{:.4f}".format
        }
    )
)
```

```python
# CELL 12 — AUTOMATICALLY SELECT BEST MODEL BY TEST ACCURACY

best_model_name = results_df.iloc[0]["Model"]
best_accuracy = results_df.iloc[0]["Accuracy"]

best_pipeline = trained_pipelines[
    best_model_name
]

print("=" * 70)
print("STEP 12 - BEST MODEL")
print("=" * 70)

print("Best Model:", best_model_name)
print("Accuracy:", f"{best_accuracy:.2%}")
```

```python
# CELL 13 — FINAL TEST RESULTS

final_predictions = best_pipeline.predict(X_test)

final_probabilities = best_pipeline.predict_proba(
    X_test
)[:, 1]

final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

final_precision = precision_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_predictions,
    zero_division=0
)

final_roc_auc = roc_auc_score(
    y_test,
    final_probabilities
)

print("=" * 70)
print("STEP 13 - FINAL TEST RESULTS")
print("=" * 70)

print("Model     :", best_model_name)
print("Accuracy  :", f"{final_accuracy:.2%}")
print("Precision :", f"{final_precision:.2%}")
print("Recall    :", f"{final_recall:.2%}")
print("F1 Score  :", f"{final_f1:.2%}")
print("ROC-AUC   :", f"{final_roc_auc:.2%}")
```

```python
# CELL 14 — CLASSIFICATION REPORT

print("=" * 70)
print("STEP 14 - CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        final_predictions,
        target_names=["Not Fraud", "Fraud"],
        zero_division=0
    )
)
```

```python
# CELL 15 — CONFUSION MATRIX

print("=" * 70)
print("STEP 15 - CONFUSION MATRIX")
print("=" * 70)

cm = confusion_matrix(
    y_test,
    final_predictions
)

print(cm)
```

```python
# CELL 16 — ACCURACY CHECK

print("=" * 70)
print("STEP 16 - ACCURACY CHECK")
print("=" * 70)

print(
    f"Genuine test accuracy: {final_accuracy:.2%}"
)

if final_accuracy >= 0.80:
    print("Test accuracy is 80% or higher.")
else:
    print("Test accuracy is below 80%.")

print("No artificial accuracy adjustment was used.")
```

```python
# CELL 17 — SAVE FINAL MODEL

MODEL_PATH = Path("vehicle_fraud_final_model.pkl")

joblib.dump(
    best_pipeline,
    MODEL_PATH
)

print("=" * 70)
print("STEP 17 - MODEL SAVED")
print("=" * 70)

print("Saved:", MODEL_PATH.resolve())
print("Final model:", best_model_name)
print("Final accuracy:", f"{final_accuracy:.2%}")
```

```python
# CELL 18 — VERIFY SAVED MODEL

loaded_model = joblib.load(
    MODEL_PATH
)

verification_predictions = loaded_model.predict(
    X_test
)

verification_accuracy = accuracy_score(
    y_test,
    verification_predictions
)

print("=" * 70)
print("STEP 18 - SAVED MODEL VERIFICATION")
print("=" * 70)

print("Loaded model:", type(loaded_model).__name__)
print("Verification accuracy:", f"{verification_accuracy:.2%}")
print("Model file exists:", MODEL_PATH.exists())
```
